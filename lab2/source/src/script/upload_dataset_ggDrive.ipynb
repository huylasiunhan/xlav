```python
import os, random, shutil, zipfile
from pathlib import Path
from google.colab import drive
import kagglehub

# 1. Kết nối Google Drive
drive.mount('/content/drive')

# 2. Tải dataset trực tiếp về ổ tạm của Colab (không tốn dung lượng máy bạn)
raw_path = Path(kagglehub.dataset_download("arunrk7/surface-crack-detection"))
print("Downloaded to:", raw_path)

# 3. Cấu hình số lượng ảnh mỗi lớp
# - Đặt SAMPLES_PER_CLASS = 2000 -> tổng 4.000 ảnh (~25 MB, train rất nhanh, vẫn đủ thử thách)
# - Đặt SAMPLES_PER_CLASS = None -> lấy toàn bộ 40.000 ảnh (~235 MB)
SAMPLES_PER_CLASS = 2000
SEED = 42

drive_dir = Path("/content/drive/MyDrive/XLAV_Lab2")
drive_dir.mkdir(parents=True, exist_ok=True)
zip_path = drive_dir / "surface_crack_dataset.zip"

random.seed(SEED)
with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_STORED) as zf:
    for cls in ["Positive", "Negative"]:
        cls_dir = raw_path / cls
        imgs = sorted([p for p in cls_dir.iterdir() if p.suffix.lower() in {".jpg", ".jpeg", ".png"}])
        if SAMPLES_PER_CLASS is not None:
            imgs = random.sample(imgs, SAMPLES_PER_CLASS)
        for img_path in imgs:
            zf.write(img_path, arcname=f"{cls}/{img_path.name}")
        print(f"Class {cls}: packed {len(imgs)} images")

print(f"Done! Saved backup zip at: {zip_path} ({zip_path.stat().st_size / 1e6:.1f} MB)")
```
